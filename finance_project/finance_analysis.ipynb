```python

num_cols=df1.select_dtypes(include=["int64","float64"]).columns.tolist()
num_cols = [c for c in num_cols if c not in ['id', 'member_id']]
for col in num_cols:
    df1[col]=pd.to_numeric(df1[col],errors="coerce")
    Q1=df1[col].quantile(0.25)
    Q3=df1[col].quantile(0.75)
    IQR=Q3-Q1
    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR
    df1[col]=df1[col].apply(lambda x:lower_bound if x < lower_bound else(upper_bound if x > upper_bound else x))
```

```python
def emp_num(x):
    if  x == "Unknown":
        return 0
    x = x.strip()
    if x == "< 1 year":
        return 0.5
    x = x.replace("+", "")
    x = x.replace(" years", "")
    x = x.replace(" year", "")

    try:
        return int(x)  # integers for everything else
    except:
        return 0

df1['emp_len_new'] = df1['emp_length'].apply(emp_num)

df1['emp_length_display'] = df1['emp_len_new'].apply(lambda x: '0.5' if x == 0.5 else str(int(x)))

```
